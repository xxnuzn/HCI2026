```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/girl.jpg')
gImg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gImg)
```

```python
# gImg.shape[0]과 gImg.shape[1] 사이에 쉼표(,) 추가
gPlus = np.zeros((gImg.shape[0], gImg.shape[1]), dtype=np.ubyte)
```

```python
import numpy as np

val = 140

# gImg의 높이(h)와 너비(w)만큼 2중 루프 돌기
for h in range(gImg.shape[0]):
    for w in range(gImg.shape[1]):
        plusVal = np.int32(gImg[h][w]) + val  # 현재 픽셀 값에 100을 더함

        if plusVal > 255:           # 255를 초과하면
            plusVal = 255           # 255로 고정 (포화 연산)

        gPlus[h][w] = plusVal       # 결과 배열에 저장

cv2_imshow(gPlus)                  # 결과 이미지 출력 (Google Colab용)
```

```python
# gImg.shape[0]과 gImg.shape[1] 사이에 쉼표(,) 추가
gOrigin = np.zeros((gImg.shape[0], gImg.shape[1]), dtype=np.ubyte)
```

```python
import numpy as np

# 주석 수정: val = 140 (140을 뺌)
val = 140

for h in range(gImg.shape[0]):
    for w in range(gImg.shape[1]):
        plusVal = np.int32(gImg[h][w]) - val

        # 1. 상한선(255 초과) 처리
        if plusVal > 255:
            plusVal = 255
        # 2. 하한선(0 미만) 처리 (★핵심 누락 부분!)
        elif plusVal < 0:
            plusVal = 0

        gOrigin[h][w] = plusVal

cv2_imshow(gOrigin)
```
