```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
```

```python
def saturation(value):
  if value > 255:
    value = 255
  elif value < 0:
    value = 0
  return value
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/aero2.bmp')
gImg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gImg)
```

```python
# gImg.shape[0]과 gImg.shape[1] 사이에 쉼표(,) 추가
gOut = np.zeros((gImg.shape[0], gImg.shape[1]), dtype=np.ubyte)
```

```python
min = 256
max = 0

for h in range(gImg.shape[0]):
    for w in range(gImg.shape[1]):

        if gImg[h, w] < min:
            min = gImg[h, w]

        if gImg[h, w] > max:
            max = gImg[h, w]

print(min)
print(max)
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/girl.jpg')
gImg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gImg)
```

```python
gMul = np.zeros((gImg.shape[0], gImg.shape[1]), dtype=np.ubyte)
```

```python
mulVal = 2
for h in range(gImg.shape[0]):
    for w in range(gImg.shape[1]):
        val = np.int32(gImg[h, w]) * mulVal
        gMul[h, w] = saturation(val)

cv2_imshow(gMul)
```

```python
gDiv = np.zeros((gImg.shape[0], gImg.shape[1]), dtype=np.ubyte)
```

```python
divVal = 2
for h in range(gImg.shape[0]):
    for w in range(gImg.shape[1]):
        val = np.int32(gMul[h, w]) / divVal
        gDiv[h, w] = saturation(val)

cv2_imshow(gDiv)
```
